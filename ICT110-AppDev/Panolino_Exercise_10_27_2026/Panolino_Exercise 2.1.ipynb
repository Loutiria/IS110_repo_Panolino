Load to a dataframe the file `GDP(PPP)_LifeExpectancy.xlsx` in the folder `Datasets`.
- Don't load the column Code which contains the full name of the country
- Use the following column names: 'Country', 'Life_Exp', 'GDP', 'Population', 'Inc_Class'
- Encode invalid values to `np.nan`

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
import pandas as pd
import numpy as np
import os

# Dynamic search using os.walk to locate the file in Google Drive
target_file = 'GDP(PPP)_LifeExpectancy.xlsx'
file_path = None

for root, dirs, files in os.walk('/content/drive/MyDrive'):
    if target_file in files:
        file_path = os.path.join(root, target_file)
        break

if not file_path:
    raise FileNotFoundError(f"Could not find {target_file} inside your Google Drive. Please ensure the file exists.")

print(f"Found file at: {file_path}")

# Load the Excel file:
# - We read columns 0, 2, 3, 4, 5 to skip the 'Code' column (which is at index 1)
# - Assign names: 'Country', 'Life_Exp', 'GDP', 'Population', 'Inc_Class'
# - Map invalid strings like '..' to np.nan using na_values
df_gdp = pd.read_excel(
    file_path,
    usecols=[0, 2, 3, 4, 5],
    names=['Country', 'Life_Exp', 'GDP', 'Population', 'Inc_Class'],
    na_values=['..']
)

# Display the loaded DataFrame
display(df_gdp.head())
```
